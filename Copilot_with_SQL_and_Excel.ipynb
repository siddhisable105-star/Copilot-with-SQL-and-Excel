Question 1: Create a SalesData table and insert 10,000 rows of random data.

Objective: To create a table named SalesData and populate it with 10,000 records using SQL.

SQL Query

CREATE TABLE SalesData (
    CustomerID INT,
    Name VARCHAR(100),
    Age INT,
    City VARCHAR(50),
    PurchaseAmount DECIMAL(10,2),
    PurchaseDate DATE
);

Insert 10,000 Random Records

WITH Numbers AS (
    SELECT TOP (10000)
        ROW_NUMBER() OVER (ORDER BY (SELECT NULL)) AS ID
    FROM sys.all_objects a
    CROSS JOIN sys.all_objects b
)
INSERT INTO SalesData
    (CustomerID, Name, Age, City, PurchaseAmount, PurchaseDate)
SELECT
    ID,
    CONCAT('Customer', ID),
    18 + ABS(CHECKSUM(NEWID())) % 48,
    CASE ABS(CHECKSUM(NEWID())) % 5
        WHEN 0 THEN 'Pune'
        WHEN 1 THEN 'Mumbai'
        WHEN 2 THEN 'Lonavala'
        WHEN 3 THEN 'Nashik'
        ELSE 'Nagpur'
    END,
    CAST(100 + ABS(CHECKSUM(NEWID())) % 9900 AS DECIMAL(10,2)),
    DATEADD(
        DAY,
        - (ABS(CHECKSUM(NEWID())) % 365),
        CAST(GETDATE() AS DATE)
    )
FROM Numbers;

Verify the Data

SELECT COUNT(*) AS TotalRecords
FROM SalesData;

SELECT TOP 10 *
FROM SalesData;

Expected output: The table contains 10,000 customer records with randomly generated ages, cities, purchase amounts, and purchase dates.

Result: The SalesData table was created and populated with 10,000 records.

Question 2: Find total sales per city and the top 5 cities by revenue.

Objective: To calculate total sales for each city and identify the five cities generating the highest revenue.

A. Total Sales Per City

SELECT
    City,
    SUM(PurchaseAmount) AS TotalSales
FROM SalesData
GROUP BY City
ORDER BY TotalSales DESC;

Explanation:

SUM() calculates total purchase amounts.
GROUP BY City groups the records by city.
ORDER BY TotalSales DESC displays the cities from highest to lowest sales.
B. Top 5 Cities by Revenue

SELECT TOP 5
    City,
    SUM(PurchaseAmount) AS TotalRevenue
FROM SalesData
GROUP BY City
ORDER BY TotalRevenue DESC;

SELECT TOP 5
    City,
    SUM(PurchaseAmount) AS TotalRevenue
FROM SalesData
GROUP BY City
ORDER BY TotalRevenue DESC;

Explanation: The query calculates total revenue for every city and returns the five cities with the highest revenue.

Result: The queries identify total sales by city and rank cities according to revenue. The actual ranking depends on the generated data.

Question 3: Find customers with purchases above average.

Objective: To identify customers whose purchase amounts are greater than the average purchase amount across all records.

SELECT
    CustomerID,
    Name,
    City,
    PurchaseAmount
FROM SalesData
WHERE PurchaseAmount > (
    SELECT AVG(PurchaseAmount)
    FROM SalesData
)
ORDER BY PurchaseAmount DESC;

Explanation
The inner query calculates the average purchase amount.
The outer query selects customers whose purchase amounts exceed that average.
The results are sorted from highest to lowest purchase amount.
Example

Suppose the average purchase amount is ₹5,000

| Customer  | Purchase Amount | Result        |
| --------- | --------------: | ------------- |
| Customer1 |          ₹7,000 | Above average |
| Customer2 |          ₹3,000 | Below average |
| Customer3 |          ₹8,000 | Above average |


Result: The query displays all records with purchase amounts greater than the overall average.

Question 8: Use Copilot to generate a new calculated column named Total Sales.

Objective: To create a calculated column that determines the total sales value for each record.

Example

Suppose the dataset contains these columns:

Quantity
Unit Price

The Total Sales column can be calculated by multiplying quantity by unit price.

Excel Formula

=[@Quantity]*[@[Unit Price]]

This structured-reference formula works when the data is formatted as an Excel table and the column names match exactly.

Alternatively, if Quantity is in column B and Unit Price is in column C, enter the following formula in row 2:

=B2*C2

| Product B |        3 |        500 |       1,500 |

Explanation

The Total Sales column calculates the value of each product's sales using the quantity purchased and the price per unit.
